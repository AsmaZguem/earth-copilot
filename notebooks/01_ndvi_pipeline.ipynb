# Step 1 — Sentinel-2 → NDVI → PNG

Area: Lac Tremblant (Quebec). Expect negative NDVI on the lake and high NDVI on the forest.

```python
from earth_copilot.geo.indices import bands_for, compute_index
from earth_copilot.geo.render import save_png
from earth_copilot.geo.stac import bbox_area_km2, load_reflectance, search_scenes

bbox = (-74.62, 46.18, -74.54, 46.26)
print(f"{bbox_area_km2(bbox):.1f} km²")
```

```python
scenes = search_scenes(bbox, "2024-06-01", "2024-08-31")
for s in scenes[:5]:
    print(s.id, s.datetime.date(), f"{s.properties['eo:cloud_cover']:.1f}%")
item = scenes[0]
```

```python
ds = load_reflectance(item, bands_for("NDVI"), bbox)
ndvi = compute_index("NDVI", ds)
ndvi.plot(cmap="RdYlGn", vmin=-1, vmax=1, figsize=(7, 7))
```

```python
ndvi.plot.hist(bins=50)
```

```python
corners = save_png(ndvi, "../outputs/ndvi_notebook.png")
corners
```
